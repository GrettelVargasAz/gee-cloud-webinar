## 1. Cargar las provincias de Costa Rica

Se utilizará la colección FAO GAUL 2015 nivel 1 , que contiene divisiones administrativas de primer nivel.

```javascript
var provincias = ee.FeatureCollection(
  'FAO/GAUL/2015/level1'
).filter(
  ee.Filter.eq('ADM0_NAME', 'Costa Rica')
);

print('Provincias de Costa Rica:', provincias);
print('Cantidad de provincias:', provincias.size());
```

> **Práctica:** 🧪 Práctica: ejecute el código y confirme en la Consola cuántas provincias contiene la colección.

## 2. Explorar los atributos

Antes de filtrar una provincia es necesario identificar los nombres de las propiedades disponibles.

```javascript
var primeraProvincia = provincias.first();

print('Primera provincia:', primeraProvincia);

print(
  'Propiedades disponibles:',
  primeraProvincia.propertyNames()
);
```

> **Práctica:** 🧪 Práctica: abra el primer elemento en la Consola e identifique las propiedades ADM0_NAME y ADM1_NAME .

## 3. Visualizar la colección

La colección completa se centra y se añade al mapa con borde azul y relleno semitransparente.

```javascript
Map.centerObject(provincias, 7);
Map.setOptions('HYBRID');

Map.addLayer(
  provincias,
  {
    color: '005ea8',
    fillColor: '005ea822'
  },
  'Provincias de Costa Rica'
);
```

> **Práctica:** 🧪 Práctica: cambie el color del borde y pruebe un nivel de acercamiento diferente.

## 4. Seleccionar una provincia

Se utiliza una variable para elegir la provincia y un filtro para obtener únicamente su polígono.

```javascript
var nombreProvincia = 'Guanacaste';

var provinciaSeleccionada = provincias.filter(
  ee.Filter.eq('ADM1_NAME', nombreProvincia)
);

print(
  'Provincia seleccionada:',
  provinciaSeleccionada
);

Map.addLayer(
  provinciaSeleccionada,
  {
    color: 'f9a825',
    fillColor: 'f9a82544'
  },
  nombreProvincia
);

Map.centerObject(provinciaSeleccionada, 8);
```

> **Práctica:** 🧪 Práctica: cambie 'Guanacaste' por otra provincia de Costa Rica.

## 5. Obtener la geometría

La colección filtrada incluye atributos y geometría. Con geometry() se obtiene únicamente su forma espacial.

```javascript
var geometriaProvincia =
  provinciaSeleccionada.geometry();

print(
  'Geometría de la provincia:',
  geometriaProvincia
);
```

> **Nota:** La geometría obtenida se utilizará en las siguientes operaciones.

## 6. Calcular el centroide

El centroide representa un punto central calculado a partir de la geometría de la provincia.

```javascript
var centroideProvincia =
  geometriaProvincia.centroid();

Map.addLayer(
  centroideProvincia,
  {color: 'ff0000'},
  'Centroide'
);
```

> **Práctica:** 🧪 Práctica: use Inspector para revisar las coordenadas aproximadas del centroide.

## 7. Crear un buffer

El buffer crea una zona alrededor de una geometría. Google Earth Engine utiliza metros como unidad.

```javascript
var bufferProvincia =
  centroideProvincia.buffer(25000);

Map.addLayer(
  bufferProvincia,
  {
    color: '2e8b57',
    fillColor: '2e8b5733'
  },
  'Buffer de 25 km'
);
```

> **Práctica:** 🧪 Práctica: cambie el buffer de 25 km a 50 km utilizando 50000 .

## 8. Obtener el rectángulo envolvente

bounds() genera el rectángulo mínimo que contiene toda la geometría seleccionada.

```javascript
var limitesProvincia =
  geometriaProvincia.bounds();

Map.addLayer(
  limitesProvincia,
  {
    color: '7b2cbf',
    fillColor: '7b2cbf11'
  },
  'Rectángulo envolvente'
);
```

> **Práctica:** 🧪 Práctica: active y desactive la provincia y el rectángulo para comparar sus formas.

## 9. Unir todas las provincias

union() combina las geometrías de una colección y permite representar Costa Rica como una sola entidad.

```javascript
var costaRicaUnida = provincias.union();

print(
  'Costa Rica como una sola geometría:',
  costaRicaUnida
);

Map.addLayer(
  costaRicaUnida,
  {
    color: '1565c0',
    fillColor: '1565c011'
  },
  'Costa Rica unida',
  false
);
```

> **Nota:** El último valor false hace que la capa se agregue desactivada inicialmente.

## 10. Definir el área de interés

El área de interés, o AOI, es la geometría que se utilizará para filtrar imágenes, recortar resultados y realizar exportaciones.

```javascript
var areaInteres =
  provinciaSeleccionada.geometry();

print(
  'Área de interés:',
  areaInteres
);
```

> **Nota:** Importante: esta variable se reutilizará en el Script 03 para filtrar imágenes satelitales.

## 11. Comparar mapas base

El Code Editor dispone de cuatro mapas base nativos. El primer argumento define el estilo inicial y la lista final controla las opciones disponibles.

```javascript
Map.setOptions(
  'HYBRID',
  {},
  [
    'ROADMAP',
    'SATELLITE',
    'TERRAIN',
    'HYBRID'
  ]
);
```

> **Práctica:** 🧪 Práctica: compare la visibilidad de las geometrías sobre cada mapa base.

## 12. Construir un fondo topográfico

Además de los mapas base nativos, una imagen de Earth Engine puede utilizarse como fondo cartográfico temático. En este ejemplo se usa el modelo digital de elevación SRTM para crear un sombreado del relieve.

```javascript
// Cargar el modelo digital de elevación.
var dem = ee.Image('USGS/SRTMGL1_003')
  .select('elevation')
  .clip(areaInteres);

// Crear el sombreado del relieve.
var sombreado = ee.Terrain.hillshade(dem);

// Añadir el sombreado como fondo temático.
Map.addLayer(
  sombreado,
  {
    min: 0,
    max: 255,
    palette: ['202020', '707070', 'ffffff']
  },
  'Fondo topográfico',
  false,
  0.70
);

// Añadir nuevamente la provincia encima.
Map.addLayer(
  provinciaSeleccionada.style({
    color: '005ea8',
    fillColor: '00000000',
    width: 3
  }),
  {},
  'Provincia sobre fondo topográfico',
  false
);
```

> **Nota:** Diferencia importante: ROADMAP , SATELLITE , TERRAIN y HYBRID son mapas base nativos. El sombreado SRTM es una capa temática añadida con Map.addLayer() .

> **Práctica:** 🧪 Práctica: active el fondo topográfico y la provincia desde Layers . Cambie la opacidad de 0.70 a 0.50 y compare el resultado.

## 💻 Código completo

Abra esta sección para consultar o copiar el script íntegro.

```javascript
// ================================================================
// CURSO INTRODUCTORIO GOOGLE EARTH ENGINE
// SCRIPT 02 — DATOS GEOGRÁFICOS Y GEOMETRÍAS
// Autora: Grettel Vargas Azofeifa
// ================================================================

// 1. CARGAR LOS LÍMITES ADMINISTRATIVOS DE COSTA RICA
var provincias = ee.FeatureCollection(
  'FAO/GAUL/2015/level1'
).filter(
  ee.Filter.eq('ADM0_NAME', 'Costa Rica')
);

print('Provincias de Costa Rica:', provincias);
print('Cantidad de provincias:', provincias.size());

// 2. EXPLORAR LA COLECCIÓN
var primeraProvincia = provincias.first();

print('Primera provincia:', primeraProvincia);
print(
  'Propiedades disponibles:',
  primeraProvincia.propertyNames()
);

// 3. VISUALIZAR LAS PROVINCIAS
Map.centerObject(provincias, 7);
Map.setOptions('HYBRID');

Map.addLayer(
  provincias,
  {
    color: '005ea8',
    fillColor: '005ea822'
  },
  'Provincias de Costa Rica'
);

// 4. SELECCIONAR UNA PROVINCIA
var nombreProvincia = 'Guanacaste';

var provinciaSeleccionada = provincias.filter(
  ee.Filter.eq('ADM1_NAME', nombreProvincia)
);

print(
  'Provincia seleccionada:',
  provinciaSeleccionada
);

Map.addLayer(
  provinciaSeleccionada,
  {
    color: 'f9a825',
    fillColor: 'f9a82544'
  },
  nombreProvincia
);

Map.centerObject(provinciaSeleccionada, 8);

// 5. OBTENER LA GEOMETRÍA
var geometriaProvincia = provinciaSeleccionada.geometry();

print(
  'Geometría de la provincia:',
  geometriaProvincia
);

// 6. CALCULAR EL CENTROIDE
var centroideProvincia = geometriaProvincia.centroid();

Map.addLayer(
  centroideProvincia,
  {
    color: 'ff0000'
  },
  'Centroide'
);

// 7. CREAR UN BUFFER DE 25 KM
var bufferProvincia = centroideProvincia.buffer(25000);

Map.addLayer(
  bufferProvincia,
  {
    color: '2e8b57',
    fillColor: '2e8b5733'
  },
  'Buffer de 25 km'
);

// 8. OBTENER EL RECTÁNGULO ENVOLVENTE
var limitesProvincia = geometriaProvincia.bounds();

Map.addLayer(
  limitesProvincia,
  {
    color: '7b2cbf',
    fillColor: '7b2cbf11'
  },
  'Rectángulo envolvente'
);

// 9. UNIR TODAS LAS PROVINCIAS
var costaRicaUnida = provincias.union();

print(
  'Costa Rica como una sola geometría:',
  costaRicaUnida
);

Map.addLayer(
  costaRicaUnida,
  {
    color: '1565c0',
    fillColor: '1565c011'
  },
  'Costa Rica unida',
  false
);

// 10. CREAR EL ÁREA DE INTERÉS
var areaInteres = provinciaSeleccionada.geometry();

print('Área de interés:', areaInteres);

// 11. COMPARAR MAPAS BASE
Map.setOptions(
  'HYBRID',
  {},
  [
    'ROADMAP',
    'SATELLITE',
    'TERRAIN',
    'HYBRID'
  ]
);

// 12. CREAR UN FONDO TOPOGRÁFICO
var dem = ee.Image('USGS/SRTMGL1_003')
  .select('elevation')
  .clip(areaInteres);

var sombreado = ee.Terrain.hillshade(dem);

Map.addLayer(
  sombreado,
  {
    min: 0,
    max: 255,
    palette: ['202020', '707070', 'ffffff']
  },
  'Fondo topográfico',
  false,
  0.70
);

Map.addLayer(
  provinciaSeleccionada.style({
    color: '005ea8',
    fillColor: '00000000',
    width: 3
  }),
  {},
  'Provincia sobre fondo topográfico',
  false
);

// 13. RESUMEN
print(
  'Resumen:',
  'Se cargaron las provincias de Costa Rica, se seleccionó ' +
  'una provincia, se realizaron operaciones geométricas ' +
  'y se construyó un fondo topográfico con SRTM.'
);
```
