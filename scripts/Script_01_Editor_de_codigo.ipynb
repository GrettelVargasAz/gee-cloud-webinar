## 2. Mostrar información en la Consola

print() muestra mensajes u objetos en la pestaña Consola .

```javascript
print('¡Bienvenidos al webinar!');
print('Primer script ejecutado correctamente.');

print('Estoy aprendiendo Google Earth Engine.');
```

## 3. Crear variables

Una variable guarda un valor que puede reutilizarse. En estos ejemplos introductorios se utiliza la palabra var .

```javascript
var pais = 'Costa Rica';
var anio = 2026;
var nubosidadMaxima = 20;
var cursoPosterior = 'SEPAL';

print('País:', pais);
print('Año:', anio);
print('Nubosidad máxima:', nubosidadMaxima);
print('Continuidad de la capacitación:', cursoPosterior);
```

> **Práctica:** 🧪 Práctica: cambie el país o el porcentaje máximo de nubosidad.

## 4. Crear una lista

Una lista contiene varios valores ordenados. Más adelante se utilizarán listas para seleccionar bandas.

```javascript
var bandasColorNatural = [
  'B4',
  'B3',
  'B8'
];

print('Bandas de color natural:', bandasColorNatural);
print('Primera banda:', bandasColorNatural[0]);
```

> **Práctica:** 🧪 Práctica: agregue 'B8' al final de la lista y revise la Consola.

## 5. Crear un objeto de parámetros

Un objeto organiza pares propiedad: valor . Se utilizará para configurar visualizaciones y exportaciones.

```javascript
var visualizacion = {
  bands: bandasColorNatural,
  min: 0,
  max: 3000
};

print('Parámetros de visualización:', visualizacion);
print('Valor máximo:', visualizacion.max);
```

> **Práctica:** 🧪 Práctica: cambie max de 3000 a 2500.

## 6. Centrar el mapa

Map.setCenter(longitud, latitud, zoom) centra el mapa en unas coordenadas y define el nivel de acercamiento.

```javascript
Map.setCenter(
  -84.10,
  9.93,
  9
);
```

> **Práctica:** 🧪 Práctica: cambie el nivel de zoom de 7 a 5 o 9.

## 7. Cambiar el mapa base

Map.setOptions() permite cambiar el fondo cartográfico.

```javascript
Map.setOptions('HYBRID');
```

> **Práctica:** 🧪 Práctica: pruebe 'ROADMAP' , 'TERRAIN' o 'HYBRID' .
